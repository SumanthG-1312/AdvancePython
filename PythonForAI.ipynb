# Guessing Game Using Loops and Conditionnal Statements

```python
# import random module for generating random value

jackpot = random.randint(1,100) # Store the random guess value in jackpot variable

guess = int(input("Guess the Number")) # take a Input from user

count = 1 # count Number of Attempts

while guess != jackpot: # while guess not equal to the jackpot (untill the guess is equal perform this loop , if both are eual then skip this loop and jump to else statement)
  if guess > jackpot: # check if guess value greather than the jackpot
    print("Guess Lover") # then print Guess lower value(the guessed value is higher than the jackpot value)
  else :
    print("Guess Higher") # Guess higher value (the guessed value is lower than the jackpot value)
  guess = int(input("Guess Again")) # Give a chance to user by allowing him to enter a new value and check if the value meets the jackpot value , if not the usser will nter values untill he guesses the corect value
  count+=1 # count no of times user entered new values

else:
  print("Correct Guess") # if guess == jackpot , then the code skip the while statement and jums here directly
  print("No Of Attempts",count) # print the stored Attempts of the User.
```

```python
# Loop through numbers from 2 to 10 (we start from 2 because 1 is not a prime number)
for i in range(2, 11):

    # Inner loop checks if 'i' has any divisor other than 1 and itself
    # We check from 2 up to i-1
    for j in range(2, i):

        # If 'i' is divisible by 'j', then it is NOT a prime number
        if i % j == 0:
            # Break immediately because we found a divisor
            break

    # This 'else' belongs to the FOR loop (not the if statement)
    # It executes only if the loop was NOT terminated by 'break'
    else:
        # If no divisor was found, then 'i' is a prime number
        print(i)
```

```python
# Continue

for i in range(1,11):
  if i == 7 :
    continue # Continue statement is used when a programmer wants to skip any Value/Product/any section that should not be included in the list of Values/Products
  print(i)
```

# Set
A Set is an unordered collection of items. Every set element is unique(no duplicates) and must be immutable (cannot be changes).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.......

Characterstics
  * Unordered
  * Mutable
  * No Duplicates
  * Can't contain mutable data types

```python
# empty set

s = {}
print(s)

# hemo genious

s1 = {1,2,3,4,5,6,7}
print(s1)

# hetro genious

s2 = {1,4.5,"Hello",True}  # Output does not return (True) because in python True is defined as 1 , we already have 1 and set doesnot accept duplicates right!
print(s2) # output is not in order ass we given , because sets are unordered , internally it  lies on hashing, hashing decides which item should be placed at which position

# set using type conversion

s3 = set([1,2,3,4,5,7])
print(s3)

# Duplicates not allowed

s4 = {1,2,3,4,5,6,7,7,7} # Automatically removes duplicate values
print(s4)

# set can't have mutable items

s5 = {1,2,[3,4]}
```

A Higher‑Order Function (HOF) is just a function that works with other functions.

* It can take a function as input (argument).

* Or it can give back a function as output (return value).





```python
def square(x):
  return x**2
def transform(f,l): # here inplace of f, we will pass function while function calling
  output = []
  for i in l :
    output.append(f(i))
  print(output)

L = [1,2,3,4,5,6,7]
transform(square,L) # passing function to as argument to a function. This is called igher order funcction.
```

####  Magic Method

*   __add__
*   __sub__
*   __mul__
*   __truediv__ for division

```python
class expontential:
  def __init__(self,x,y):
    self.num = x
    self.den = y

  # Addition

  def __add__(obj1,obj2):
    new_num = obj1.num*obj2.den + obj2.num*obj1.den
    new_den = obj1.den*obj2.den

    return '{}/{}'.format(new_num,new_den)
    # Subtraction

  def __sub__(obj1,obj2):
    new_num = obj1.num*obj2.den - obj2.num*obj1.den
    new_den = obj1.den*obj2.den

    return '{}/{}'.format(new_num,new_den)

  # Multiplication

  def __mul__(obj1,obj2):
    new_num = obj1.num*obj2.num
    new_den = obj1.den*obj2.den

    return '{}/{}'.format(new_num,new_den)

  # Division

  def __truediv__(obj1,obj2):
    new_num = obj1.num*obj2.den
    new_den = obj1.den*obj2.num

    return '{}/{}'.format(new_num,new_den)

```

### Build 2D Co-Ordinates usinng OOP

```python
import math

class point :
  def __init__(self,x,y):
    self.x_cord = x
    self.y_cord = y

  def __str__(self):
    return '<{},{}>'.format(self.x_cord, self.y_cord)

  def euclidean_distance(self,other):
    distance = math.sqrt((self.x_cord-other.x_cord)**2 + (self.y_cord-other.y_cord)**2) # instead of math we can can return distance**0.5
    return distance

  def distance_from_orgin(self):
    return self.euclidean_distance(point(0,0)) # under root(x**2 + y**2)
```

```python
cord1 = point(6,15)
cord2 = point(13,7)
print(cord1)
print(cord2)

print(cord1.euclidean_distance(cord2))

print(cord2.distance_from_orgin())
```

```python
class line:
  def __init__(self,A,B,C):
    self.A = A
    self.B = B
    self.C = C

  def __str__(self):
    return '{}x + {}y + {} = 0'.format(self.A,self.B,self.C)

  def point_on_line(self,point):
    if self.A*point.x_cord + self.B*point.y_cord + self.C == 0:
      return 'lies on the line'
    else:
      return 'does not lie on the line'

  def shortest_distance(self,point):
    return abs(self.A*point.x_cord + self.B*point.y_cord + self.C)/(self.A**2 + self.B**2)**0.5
```

```python
l1 = line(13,7,-3)
p1 = point(7,8)

l1.point_on_line(p1)

l1.shortest_distance(p1)
```

# Encapsulation

1.  Hide variables and internal logic so they can’t be directly accessed or modified from outside the class.
2.   Expose controlled access through getter and setter methods (or properties in Python).
3.   This way, you decide what’s allowed and what’s not, instead of letting external code mess with your object’s state.

________________________________________________________________________________

Private Variables in Python:

    Declared as __variable.

    Internally, Python performs name mangling → _ClassName__variable.

This prevents accidental access but still allows controlled access if needed.

* Getter Methods: Safely retrieve private data.

* Setter Methods: Modify private data with validation logic (e.g., only integers, >0, within limits).

Benefits:

Data safety and consistency.

Flexibility to change internal implementation without breaking external code.

Cleaner, modular, and maintainable design.

--------------------------------------------------------------------------------

* Shortcut memory trick:
  * Encapsulation = Hide + Protect + Control.

```python
class ZomatoUser:
    def __init__(self, name, age, gender):
        self.__name = name      # private variable
        self.__age = age        # private variable
        self.__gender = gender  # private variable

    # Getter methods
    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

    def get_gender(self):
        return self.__gender

    # Setter methods with basic logic (no isinstance)
    def set_age(self, age):
        if age > 0:   # only positive values allowed
            self.__age = age
        else:
            print("Invalid age: must be greater than 0")

    def set_name(self, name):
        if len(name) > 0:   # name must not be empty
            self.__name = name
        else:
            print("Invalid name: cannot be empty")
```

```python
user = ZomatoUser("Sumanth", 21, "Male")

print(user.get_name())   # Sumanth
print(user.get_age())    # 21


user.set_age(-5)         # Invalid age: must be greater than 0
user.set_age(22)         # Works fine
print(user.get_age())    # 22
```

# Super()

super() returns a proxy object that represents the parent class.

It gives access to methods and properties of the parent class from within the child class.

Commonly used in inheritance to:

Call the parent’s constructor (__init__) to initialize attributes.

Call parent methods when overriding them in the child.

```python
class ZomatoUser:
    def __init__(self, name):
        self.name = name

    def order_food(self):
        print(f"{self.name} is ordering food from Zomato (normal process).")


class PremiumUser(ZomatoUser):
    def __init__(self, name, membership):
        super().__init__(name)   # call parent constructor
        self.membership = membership

    # Method overriding
    def order_food(self):
        print(f"{self.name} is ordering food with {self.membership} benefits (fast delivery + discounts).")
```

```python
user1 = ZomatoUser("Sumanth")
user2 = PremiumUser("Sumanth", "Gold Membership")

user1.order_food()
user2.order_food()
```

```python
# GrandParent Class

class Zomato:
    def __init__(self, app_name): #
        self.app_name = app_name

    def show_app(self):
        return f"Welcome to {self.app_name}"

# Parent Class

class Restaurant(Zomato):
    def __init__(self, app_name, restaurant_name):
        super().__init__(app_name) # App name is being inherited from parent class thorugh  super()
        self.restaurant_name = restaurant_name

    def show_restaurant(self):
        return f"Restaurant: {self.restaurant_name}"

# Child Class

class Dish(Restaurant):
    def __init__(self, app_name, restaurant_name, dish_name):
        super().__init__(app_name, restaurant_name) # inheriting both parent and grandparent attributes here
        self.dish_name = dish_name

    def show_dish(self):
        return f"Dish: {self.dish_name}"
```

```python
user = Dish("Zomato", "Lucky", "Biryani")
user.show_app()
user.show_restaurant()
user.show_dish()
```

```python
class ZomatoDelivery:
    def delivery_service(self):
        return "Delivery service enabled"

class ZomatoPayment:
    def payment_service(self):
        return "Payment service enabled"

# Child class inherits from both Delivery and Payment
class ZomatoApp(ZomatoDelivery, ZomatoPayment): # This child class can access both the parent class methods and attributes
    def app_features(self):
        return "Zomato App with multiple services"
```

> What Happens if Both Parents Have the Same Method?

  * If two parent classes define a method with the same name, and the child class inherits from both:

    * Python will execute the method from the parent that appears first in the child’s inheritance list.

* This order is determined by the MRO (Method Resolution Order).

```python
class ZomatoDelivery:
    def service(self):
        return "Delivery service from ZomatoDelivery"

class ZomatoPayment:
    def service(self):
        return "Payment service from ZomatoPayment"
# Both he Classes have same methods

# Child inherits from both
class ZomatoApp(ZomatoDelivery, ZomatoPayment):
    pass

z = ZomatoApp() # object created
print(z.service()) # Class A method will be called, coz it is first inherited, if Class B inherited 1st then it will be called.
```

# Abstraction

* abstraction is the concept of hiding complex details and showing only the essential features

* What is an Abstract Class?
  * An abstract class is a restricted class that cannot be used to create objects directly. It acts strictly as a template for other classes to inherit from.

```python
from abc import ABC, abstractmethod

# Abstract class
class FoodOrder(ABC):
    @abstractmethod
    def place_order(self, restaurant, item):
        pass   # only defines the rule, no logic here

# Concrete class
class ZomatoOrder(FoodOrder):
    def place_order(self, restaurant, item):
        # actual implementation
        print(f"Order placed for {item} from {restaurant}")

# Usage
order = ZomatoOrder()
order.place_order("Dominos", "Pizza")
```

# Pickling

Purpose: Pickling lets you store Python objects (like dicts, lists, or trained models) without losing their structure.

Process:

Object in memory → converted into binary format (byte stream).

Written into a .pkl file using "wb" mode → stored on your disk (SSD/HDD).

Original object in memory remains unchanged.

After program ends: Normally, objects vanish from RAM. But with a .pkl file, you can reload (unpickle) the object later.

Unpickling: Reads the binary data back into memory → reconstructs the same Python object with no data loss.

*  In short: Pickling is like freezing your object into a file, and unpickling is thawing it back into memory when you need it.

-----------------------------------------------------------------------------------------------

### Why convertion into binary only.

Why Pickling Uses Binary Format
Objects are not plain text. Python objects (dicts, lists, classes, models) have internal structure that text cannot capture.

Binary preserves exact state. Pickle encodes types, values, and relationships so they can be reconstructed precisely.

Efficiency. Binary is compact and faster to read/write compared to text formats.

Flexibility. Works for almost any Python object, not just simple strings or numbers.

Accuracy. Avoids ambiguity — text would only give a description, while binary ensures the object comes back exactly as it was.

```python
import pickle

# Original object
data = {"name": "Sumanth", "scores": [90, 85]}

# Pickling: save to file in binary format
with open("data.pkl", "wb") as f:
    pickle.dump(data, f)
```

```python
import pickle

# --- Pickling (save object) ---
data = {"name": "Sumanth", "scores": [90, 85]}
with open("data.pkl", "wb") as f:
    pickle.dump(data, f)

# --- Unpickling (load object back) ---
with open("data.pkl", "rb") as f:
    loaded_data = pickle.load(f)

print(loaded_data)        # {'name': 'Sumanth', 'scores': [90, 85]}
print(type(loaded_data))  # <class 'dict'>
```

## Google login/logout Example

```python
class SecurityError(Exception):
    def __init__(self, message):
        super().__init__(message)   # keep Exception behavior
        print(message)

    def logout(self):
        print("Logout from all Devices")


class Google:
    def __init__(self, email, password, device):
        self.email = email
        self.password = password
        self.device = device

    def login(self, email, password, device):
        if email == self.email and password == self.password:
            if device != self.device:
                raise SecurityError("Unregistered Device")
            print("Login Successful")
        else:
            raise SecurityError("Invalid credentials")
```

# LEGB RUle :

LEGB Rule
When Python looks for a variable, it follows this order:

L → Local: Inside the current function.

E → Enclosing: In outer functions (if nested).

G → Global: Defined at the top level of the file.

B → Built-in: Predefined names in Python (like len, print).

👉 If Python doesn’t find the variable in Local, it checks Enclosing, then Global, then Built-in. If still not found → NameError.

### Enclosing Scope

The enclosing scope is the scope of any outer function that wraps around an inner (nested) function.



```python
def outer() : # This is Enclosing / Non-Local Scope
  def inner() :
    print("Inner")
  inner()
  print("Outer ")

outer()
print("Main Program")

# Output:

'''Inner
Outer
Main Program'''
```

Checks in Local Scope, finds it and prints it.

```python
def outer() : # This is Enclosing / Non-Local Scope

  a = 13

  def inner() : # This becomes local
    a = 10 # a local variable in inner()
    print("Inner",a) # The interpreter checks if a is present in inner()/local, if yes it prints it.
  inner()
  print("Outer ")

outer()
print("Main Program")
```

# Decorators

* A decorator is a function that adds extra functionality to another function or class without changing its original code.

* This can happen because Python Functions are 1st Class Citizens.
  * 1st class ctizens :
    * First-class citizens in Python → Functions are treated like normal objects.

    You can:

        Assign them to variables

        Pass them as arguments

        Return them from other functions

---
* There are 2 types of deccorators in Python :    
  * Built-in Decorators --> These look like @staticmethods, @classmethods, @abstractmethod, @property, @log, @authenticate, @timeit,etc...
  #
  * user defined decorators -->  that we we programmers crete according to out needs.




Example code for a Decorator

```python
# Basically this is a wrapper function
def sample(func): # a function which is passed with anaother function as argument
  def wrapper(): # wrapper function is basically adding a extra functionality to passed function
    print("***********")
    func()
    print("***********")
  return wrapper # wrapper is returned to the sample function

def hello(): # this function is passed to the sample function
  print("Hello")

a = sample(hello) # creating a object for sample function, and becomes a function too
a() # calling a() which became a function
```

```python
def outer():
  a = 13
  def inner():
    print(a)
  return inner

a = outer()
a()
```

# Closure :
A closure function in Python is a function that remembers the variables from its enclosing scope even after that scope has finished executing.

Logical Example code for Decorator, what is the use of decorators in Real-Time

```python
# Senior developer writes backend logic
def place_order(order_id, user_id):
    print(f"Order {order_id} placed successfully for user {user_id}!")

# Junior developer adds a decorator for security
def secure_access(func):
    def wrapper(order_id, user_id):
        # Security check before executing the main function
        if user_id.startswith("ZOMATO_"):
            print("Security check passed")
            return func(order_id, user_id)
        else:
            print("Security check failed. Unauthorized user.")
    return wrapper

# Applying the decorator
@secure_access
def place_order_secure(order_id, user_id):
    print(f"Order {order_id} placed successfully for user {user_id}!")
```

```python
place_order_secure("ORDER123", "ZOMATO_USER001")
```
