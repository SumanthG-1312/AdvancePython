# What is an Iteration

* Iteration is a general term for taking each item of something, one after another. Any time you use a loop, explicit or implicit, to go over a group of items, that is iteration.

```python
# Example

for i in range(1,7):
  print(i)
```

# What is Iterator

* An Iterator is an object that allows the programmer to traverse through a sequence of data without having to store the entire data in the memory

```python
l = [x for x in range(1,88000)]

""" for i in l :
  print(i*13) """

import sys

print(sys.getsizeof(l)/1024)

x = range(1,118700) # when range() is used, then iterator is implemented, iterator picks one element and perform operations on it -> display to user and removes from memory.

""" for i in x:
  print(i*13) """

print(sys.getsizeof(x)/1024)
```

# What is Iterable

* Iterable is an object, which one can iterate over

* It generates an Iterator when passed to iter() method.

```python
# Example

L = [1,2,3]
type(L)


# L is an iterable
type(iter(L))

# iter(L) --> iterator
```

# Point to remember
* Every Iterator is also and Iterable
* Not all Iterables are Iterators

# Making of our own for loop

```python
from typing import Iterator
def own_for_loop(iterable):

  Iterator = iter(iterable)

  while True:

    try :
      print(next(Iterator))

    except StopIteration:
      break
```

```python
test = [1,2,3,4,5,6,7]

own_for_loop(test)
```

# Creating our own range() Function

```python
class my_range_function:

    def __init__(self, start, end):
        self.start = start
        self.end = end

    def __iter__(self):
        return my_range_iterator(self)


class my_range_iterator:

    def __init__(self, obj):
        self.obj = obj
        self.current = obj.start

    def __iter__(self):
        return self

    def __next__(self):

        if self.current >= self.obj.end:
            raise StopIteration

        current = self.current
        self.current += 1

        return current
```

```python
# instead of range i would be using :

for i in my_range_function(7,13):
  print(i)
```
